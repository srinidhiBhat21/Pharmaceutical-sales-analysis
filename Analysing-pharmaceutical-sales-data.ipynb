### 3 Which three drugs have the highest sales in January 2015, July 2016, September 2017.

```python
drug = ['M01AB', 'M01AE', 'N02BA', 'N02BE', 'N05B',	'N05C', 'R03','R06']
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
jan = df[df['Date'].between('2015-01-01', '2015-01-31')][drug].sum().sort_values(ascending=False).head(3)
```

```python
print(f"Highest 3 drugs of Jan are  -{jan}")
```

```python
july = df[df['Date'].between('2016-07-01', '2016-07-31')][drug].sum().sort_values(ascending=False).head(3)
```

```python
print(f"Highest 3 drugs of July are  -{july}")
```

```python
sept = df[(df['Date'] >= '2017-09-01') & (df['Date']<= '2017-09-30')][drug].sum().sort_values(ascending=False).head(3)
```

```python
print(f"Highest 3 drugs of September 2017 are  -{sept}")
```

```python
result = pd.concat([jan, july, sept], axis = 1)
```

```python
result.columns = ['January 2015', 'July 2016', 'September 2017']
print(result)
```

### Observation ; N02BE has the highest sales in jan 2015, jul 2016, sept 2017
